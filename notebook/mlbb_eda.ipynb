**Numerical Univariate Analysis**

```python
num_cols = df_copy[['Hp', 'Hp_Regen', 'Mana',
 'Mana_Regen', 'Phy_Damage', 
 'Mag_Damage', 'Phy_Defence', 
 'Mag_Defence', 'Mov_Speed',
 'Esport_Wins', 'Esport_Loss']]


summary = pd.DataFrame(
    {
        'Mean': num_cols.mean(),
        'Median': num_cols.median(),
        'Mode': num_cols.mode().iloc[0],
        'Variance': num_cols.var(),
        'Standard Deviation': num_cols.std(),
        'Minimum': num_cols.min(),
        'Maximum': num_cols.max(),
        'Range': num_cols.max() - num_cols.min(),
        'Skewness': num_cols.skew()
    }
)

display((summary).round())
```

```python
# subplot histogram

fig, axes = plt.subplots(nrows=4, ncols=3, figsize=(18,16))

axes = axes.flatten()

for i, col in enumerate(num_cols):
    
    sns.histplot(data=df_copy, x=col, kde=True, ax=axes[i])
    
    axes[i].axvline(df_copy[col].mean(), linestyle='--', label='Mean')
    axes[i].axvline(df_copy[col].median(), linestyle='-', label='Median')
    
    axes[i].set_title(f'Distribution of {col}')
    axes[i].legend()
    

# hide unused suplot
for j in range(len(num_cols), len(axes)):
    axes[j].set_visible(False)
    

plt.tight_layout()
plt.show()
```

```python
# identify outliers
def main():
    identify_outliers()
    
def identify_outliers():
    
    q1 = num_cols.quantile(0.25)
    q3 = num_cols.quantile(0.75)
    
    
    iqr = q3 - q1
    
    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr
    
    
    for col in num_cols:
        
        lower_outliers = num_cols[col][num_cols[col] < lower_limit[col]]
        
        upper_outliers = num_cols[col][num_cols[col]> upper_limit[col]]
        
        print("=====" * 13)
        print(f'{col} Column:')
        
        print(f'Lower Limit: {lower_limit[col]}')
        print(f'Lower Outliers: {lower_outliers.tolist()}')
        
        print(f'Upper Limit: {upper_limit[col]}')
        print(f'upper Outliers: {upper_outliers.tolist()}')
        
        

main()
```

```python
# identify count of outliers per numerical column
def main():

  total_outliers()

def total_outliers():

  q1 = num_cols.quantile(0.25)
  q3 = num_cols.quantile(0.75)

  iqr = q3 - q1

  lower_limit = q1 - 1.5 * iqr
  upper_limit = q3 + 1.5 * iqr

  for col in num_cols:

    total_lower_outliers = (num_cols[col] < lower_limit[col]).sum()

    total_upper_outliers = (num_cols[col] > upper_limit[col]).sum()

    print(f'Total number of lower outliers in {col}: {total_lower_outliers} \n')
    print(f'Total number of upper outliers in {col}: {total_upper_outliers} \n')


main()
```

**Data Visualization**

1. Hero Distribution by Role

```python
# group heroes by primary role
role_distribution = (
    df.groupby('Primary_Role')
    .size()
    .reset_index(name='total_count')
)

# calculate percentage
role_distribution['role_percentage'] = (
    role_distribution['total_count'] / len(df) * 100
).round()

# sort by total count
role_distribution = role_distribution.sort_values(
    by='total_count',
    ascending=False
)

role_distribution
```

```python
plt.figure(figsize=(10, 6))

# Define the order of the bars
role_order = role_distribution['Primary_Role'].tolist()

ax = sns.barplot(
    data=role_distribution,
    x='total_count',
    y='Primary_Role',
    order=role_order
)

# Add labels at the end of each bar
for i, bar in enumerate(ax.patches):
    width = bar.get_width()
    y_position = bar.get_y() + bar.get_height() / 2

    label = (
        f"{role_distribution.iloc[i]['total_count']} "
        f"({role_distribution.iloc[i]['role_percentage']:.2f}%)"
    )

    ax.text(
        width + 0.5,
        y_position,
        label,
        va='center',
        ha='left',
        fontsize=10
    )

# Formatting
ax.set_title('Hero Distribution by Primary Role', fontsize=14, pad=15)
ax.set_xlabel('Number of Heroes')
ax.set_ylabel('')
ax.set_xlim(0, role_distribution['total_count'].max() + 10)

ax.grid(axis='x', linestyle='--', alpha=0.3)
ax.set_axisbelow(True)

sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.show()
```

```python
# Group by primary role and calculate average attributes
role_stats = (
    df.groupby('Primary_Role')
    .agg(
        average_hp=('Hp', 'mean'),
        average_physical_damage=('Phy_Damage', 'mean'),
        average_physical_defense=('Phy_Defence', 'mean')
    )
    .round(2)
    .reset_index()
)

# Sort by average HP in descending order
role_stats = role_stats.sort_values(
    by='average_hp',
    ascending=False
)

role_stats
```

```python
# Define a consistent color for each role
role_colors = {
    'Fighter': '#E76F51',
    'Mage': '#457B9D',
    'Marksman': '#E9C46A',
    'Tank': '#2A9D8F',
    'Assassin': '#8E6CBB',
    'Support': '#F4A261'
}

fig, axes = plt.subplots(1, 3, figsize=(18, 7))

# Average HP
sns.barplot(
    data=role_stats,
    x='average_hp',
    y='Primary_Role',
    hue='Primary_Role',
    palette=role_colors,
    legend=False,
    ax=axes[0]
)

axes[0].set_title('Average HP by Role')
axes[0].set_xlabel('Average HP')
axes[0].set_ylabel('Primary Role')

# Average Physical Damage
sns.barplot(
    data=role_stats,
    x='average_physical_damage',
    y='Primary_Role',
    hue='Primary_Role',
    palette=role_colors,
    legend=False,
    ax=axes[1]
)

axes[1].set_title('Average Physical Damage by Role')
axes[1].set_xlabel('Average Physical Damage')
axes[1].set_ylabel('')

# Average Physical Defense
sns.barplot(
    data=role_stats,
    x='average_physical_defense',
    y='Primary_Role',
    hue='Primary_Role',
    palette=role_colors,
    legend=False,
    ax=axes[2]
)

axes[2].set_title('Average Physical Defense by Role')
axes[2].set_xlabel('Average Physical Defense')
axes[2].set_ylabel('')

plt.suptitle('Average Hero Attributes by Primary Role', fontsize=16)

sns.despine()
plt.tight_layout()
plt.show()
```

```python
# Filter heroes with at least one esports match
esport_df = df[
    (df['Esport_Wins'] + df['Esport_Loss']) > 0
].copy()

# Calculate total matches played
esport_df['total_matches_played'] = (
    esport_df['Esport_Wins'] + esport_df['Esport_Loss']
)

# Calculate win rate
esport_df['win_rate'] = (
    esport_df['Esport_Wins'] /
    esport_df['total_matches_played'] * 100
).round(2)

# Get the top 10 heroes by win rate
top_10_esport = esport_df.sort_values(
    by='win_rate',
    ascending=False
).head(10)

top_10_esport[['Name', 'Primary_Role', 'Esport_Wins', 'Esport_Loss', 'total_matches_played', 'win_rate']]
```

```python
plt.figure(figsize=(9, 5))

ax = sns.barplot(
    data=top_10_esport,
    x='win_rate',
    y='Name',
    color='steelblue',
    order=top_10_esport['Name']
)

# Add win rate and match count labels
for i, bar in enumerate(ax.patches):
    width = bar.get_width()
    y_position = bar.get_y() + bar.get_height() / 2

    label = (
        f"{top_10_esport.iloc[i]['win_rate']:.2f}% "
        f"({top_10_esport.iloc[i]['total_matches_played']} matches)"
    )

    ax.text(
        width + 0.3,
        y_position,
        label,
        va='center',
        fontsize=8
    )

ax.set_title('Top 10 Heroes by Esports Win Rate', fontsize=13, pad=12)
ax.set_xlabel('Win Rate (%)')
ax.set_ylabel('Hero')

ax.set_xlim(0, 75)
ax.grid(axis='x', linestyle='--', alpha=0.3)
ax.set_axisbelow(True)

sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.show()
```

```python
# Extract the release year from the Release_Date column
df['release_year'] = df['Release_Date'].str.extract(
    r'(\d{4})$|^(\d{4})'
).bfill(axis=1).iloc[:, 0]

# Count the number of heroes released per year
release_trends = (
    df.groupby('release_year')
    .size()
    .reset_index(name='hero_count')
)

# Sort by hero count in descending order
release_trends = release_trends.sort_values(
    by='hero_count',
    ascending=False
)

release_trends
```

```python
plt.figure(figsize=(9, 5))

ax = sns.barplot(
    data=release_trends,
    x='hero_count',
    y='release_year',
    color='steelblue'
)

# Add hero counts as labels
for bar in ax.patches:
    width = bar.get_width()
    y_position = bar.get_y() + bar.get_height() / 2

    ax.text(
        width + 0.3,
        y_position,
        f'{int(width)}',
        va='center',
        fontsize=9
    )

ax.set_title('Number of Hero Releases by Year', fontsize=13, pad=12)
ax.set_xlabel('Number of Heroes Released')
ax.set_ylabel('Release Year')

ax.set_xlim(0, release_trends['hero_count'].max() + 5)
ax.grid(axis='x', linestyle='--', alpha=0.3)
ax.set_axisbelow(True)

sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.show()
```

```python
# Calculate combined offensive and defensive attributes
df['total_damage'] = df['Phy_Damage'] + df['Mag_Damage']
df['total_defense'] = df['Phy_Defence'] + df['Mag_Defence']

# Rank heroes based on total damage and defense
# method='min' matches SQL RANK()
df['damage_rank'] = df['total_damage'].rank(
    method='min',
    ascending=False
)

df['defense_rank'] = df['total_defense'].rank(
    method='min',
    ascending=False
)

# Calculate combined ranking
df['combined_stats'] = df['damage_rank'] + df['defense_rank']

# Get the top 10 heroes
top_10_rankings = (
    df.sort_values(
        by=['combined_stats', 'damage_rank', 'defense_rank'],
        ascending=True
    )
    .head(10)
    [['Name', 'Primary_Role', 'damage_rank', 'defense_rank', 'combined_stats']]
)

top_10_rankings
```

```python
plt.figure(figsize=(9, 5))

ax = sns.barplot(
    data=top_10_rankings,
    x='combined_stats',
    y='Name',
    color='steelblue',
    order=top_10_rankings['Name']
)

# Add combined score and individual rankings as labels
for i, bar in enumerate(ax.patches):
    width = bar.get_width()
    y_position = bar.get_y() + bar.get_height() / 2

    row = top_10_rankings.iloc[i]

    label = (
        f"{int(row['combined_stats'])} "
        f"(Damage: {int(row['damage_rank'])}, "
        f"Defense: {int(row['defense_rank'])})"
    )

    ax.text(
        width + 0.3,
        y_position,
        label,
        va='center',
        fontsize=8
    )

ax.set_title('Top 10 Heroes by Combined Attribute Ranking', fontsize=13, pad=12)
ax.set_xlabel('Combined Ranking Score')
ax.set_ylabel('Hero')

ax.set_xlim(0, top_10_rankings['combined_stats'].max() + 12)
ax.grid(axis='x', linestyle='--', alpha=0.3)
ax.set_axisbelow(True)

sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.show()
```
